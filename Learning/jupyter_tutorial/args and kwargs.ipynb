```python
def myfunc(**kwargs):
    print(kwargs)
    if 'fruit' in kwargs:
        print('My fruit of choice is {}'.format(kwargs['fruit']))
    else:
        print('I did not find any fruit here')
```

```python
myfunc(car='mercedes', veggie='lettuce', fruit='Apple')
```

```python
def myfnc(name):
    new_name=''
    for letter in name:
        if name.index(letter)%2==0:
            letter=letter.upper()
            new_name=new_name+letter
        else:
            letter=letter.lower()
            new_name=new_name+letter
    return new_name
```
