```python
def splicer(mystring):
    if len(mystring)%2==0:
        return 'EVEN'
    else:
        return mystring[0]
```

```python
names=['Andy','Eve','Sally','Isar','Ishar','Ahmad']
```

```python
list(map(splicer,names))
```

```python
for item in map(splicer,names):
    print(item)
```
