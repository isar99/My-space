```python
def create_cubes(n):
    
    for x in range(n):
        yield x**3
```

```python
list(create_cubes(10))
```

```python
def gen_fibon(n):
    
    a=1
    b=1
    for i in range(n):
        yield a
        a,b = b,a+b
```

```python
for number in gen_fibon(10):
    print(number)
```

```python
def mymap(func, *seqs):
    res = []
    for args in zip(*seqs):
        res.append(func(*args))
    return res
```

```python
print(mymap(abs, [-2,-1,0,1,2]))
```

```python
print(mymap(pow,[1,2,3],[2,3,4]))
```

```python
#using list comprehension
def mymap(func, *seqs):
    return [func(*args) for args in zip(*seqs)]
```

```python
print(mymap(abs, [-2,-1,0,1,2]))
```

```python
print(mymap(pow,[1,2,3],[2,3,4]))
```

```python
print("LIST keeps DUPLICATE")
[ x+ y for x in [1,2,3] for y in [4,5,6]]
```

```python
print("SETS does not keep DUPLCATE")
{x + y for x in [1,2,3] for y in [4,5,6]}
```

```python
print("DICTIONARY neither keep DUPLICATE")
{x: y for x in [1,2,3] for y in [4,5,6]}
```
