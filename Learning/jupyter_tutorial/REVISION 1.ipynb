```python
for x in range(1,101):
    if x % 3 == 0 and x%5==0:
        print(x,' - FIZZBUZZ')
    elif x % 5 == 0:
        print(x,' - BUZZ ')
    elif x % 3 == 0:
        print(x, ' - FIZZ')
    else:
        print(x, ' - Neithr div by 3 nor 5')
```

```python
puzz = [('FIZZ', x) if x%3==0 and x%5==0 else ('BUZZ', x) if x%5==0 else ('FIZZBUZZ', x) if x%3==0 else ('NEITHER NOR', x) for x in range(1,101)]
```

```python
print("NOTE:- SYNTAX for if else if condition.")
print("SYNTAX:- [<data> if condition... else <data> if condition... else <data> if condition... else for x in <iterable>]")
```
