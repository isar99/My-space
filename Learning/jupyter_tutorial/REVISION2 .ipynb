```python
# LESSER OF TWO EVENS: Write a function that returns the lesser of two given numbers if both numbers are even,
# but returns the greater if one or both numbers are odd
```

```python
def lesser(num1,num2):
    if num1%2 == 0 and num2%2 == 0:
        if num1 < num2:
            return num1
        else:
            return num2
    if num1%2 == 1 or num2%2 == 1:
        if num1 > num2:
            return num1
        else:
            return num2
    
```

```python
lesser(1,5)
```

```python
lesser(2,3)
```

```python
#Given a list of ints, return True if the array contains a 3 next to a 3 somewhere.
def has_33(my_list):
    for num in range(len(my_list)-1):
            
        if my_list[num] == my_list[num+1] ==3:
            return True
        else:
            continue
    return False
```

```python
#PAPER DOLL: Given a string, return a string where for every character in the original there are three characters
def paper_doll(my_str):
    new_str = ''
    for letter in my_str:
        new_str = new_str + letter*3
    return new_str
```

```python
#BLACKJACK: Given three integers between 1 and 11, if their sum is less than or equal to 21, return their sum.
#If their sum exceeds 21 and there's an eleven, reduce the total sum by 10. Finally, if the sum 
#(even after adjustment) exceeds 21, return 'BUST'
def blackjack(num1,num2,num3):
    if num1+num2+num3 <= 21:
        return num1+num2+num3
    elif num1+num2+num3 >21 and (num1 ==11 or num2 ==11 or num3 ==11):
        return (num1+num2+num3) - 10
    elif num1+num2+num3 > 21:
        return 'BUST'
```

```python
class Circle:
    pi = 3.14

    # Circle gets instantiated with a radius (default is 1)
    def __init__(self, radius=1):
        self.radius = radius 
        self.area = radius * radius * Circle.pi

    # Method for resetting Radius
    def setRadius(self, new_radius):
        self.radius = new_radius
        self.area = new_radius * new_radius * self.pi

    # Method for getting Circumference
    def getCircumference(self):
        return self.radius * self.pi * 2


c = Circle()

print('Radius is: ',c.radius)
print('Area is: ',c.area)
print('Circumference is: ',c.getCircumference())
```

```python
#Fill in the Line class methods to accept coordinates as a pair of tuples and return the slope and distance of the line.

```

```python
class Line :
    def __init__(self,coor1,coor2):
        self.x1 = coor1[0]
        self.y1 = coor1[1]
        self.x2 = coor2[0]
        self.y2 = coor2[1]
    def slope(self):
        self.slope = (self.y2-self.y1)/(self.x2-self.x1)
        return self.slope
    def distance(self):
        self.distance = ((self.x2-self.x1)**2 + (self.y2-self.y1)**2)**0.5
        return self.distance
```

```python
coor1 = (3,2)
coor2 = (8,10)

li = Line(coor1,coor2)
```

```python
li.slope()
```

```python
li.distance()
```

```python
class Cylinder:
    
    pi = 3.14
    def __init__(self,height=1,radius=1):
        self.height = height
        self.radius = radius
        
    def volume(self):
        self.volume = Cylinder.pi * self.radius**2 * self.height
        return self.volume
    
    def surface_area(self):
        self.sur_area = 2* Cylinder.pi*self.radius *(self.radius +self.height)
        return self.sur_area
```

```python
c = Cylinder(2,3)
```

```python
c.volume()
```

```python
c.surface_area()
```

```python
def genfibon(n):
    """
    Generate a fibonnaci sequence up to n
    """
    a = 1
    b = 1
    for i in range(n):
        yield a
        a,b = b,a+b
```

```python
genfibon(10)
```

```python
list(genfibon(10))
```

```python
def scope_test():
    def do_local():
        spam = "local spam"

    def do_nonlocal():
        nonlocal spam
        spam = "nonlocal spam"

    def do_global():
        global spam
        spam = "global spam"

    spam = "test spam"
    do_local()
    print("After local assignment:", spam)
    do_nonlocal()
    print("After nonlocal assignment:", spam)
    do_global()
    print("After global assignment:", spam)

scope_test()
print("In global scope:", spam)
```
