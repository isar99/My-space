```python
class Account():
    
    def __init__(self,name,balance):
        self.owner= name
        self.balance= balance
        
    def deposit(self,amount):
        self.balance= self.balance + amount
        print('{} is deposited',amount)
    def withdraw(self,amount):
        if amount>self.balance:
            return f"Funds Unavailable"
        else:
            print("Withdrawl succesfull")
            return f"Current balance is {self.balance-amount}"
```

```python
acc= Account('isar',2000)
```

```python
acc.deposit(500)
```
