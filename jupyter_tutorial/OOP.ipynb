```python
class Dog():
    
    
    #CLASS OBJECT ATTRIBUTE
    #SAME FOR ANY INSTANCE OF A CLASS
    species='mammal'
    
    
    def __init__(self,breed,name):
        
        #Attributes
        #We take in the argument
        #Assign it using self.attribute_name
        self.breed=breed
        self.name=name
        
    #OPERATIONS/Actions---->Methods
    def bark(self,number):
        print('WOOF! My name is {} and number is {}'.format(self.name,number))
```

```python
my_dog=Dog('Husie','Tony')
```

```python
my_dog.bark(5)
```

```python
my_dog.species
```

```python
my_dog.name
```

```python
class Circle():
    
    #CLASS OBJECT ATTRIBUTE
    pi=3.14
    
    def __init__(self,radius=1):
        
        self.radius=radius
        self.area=self.pi*radius**2
        #WE CAN ALSO WRITE Circle.pi instead of self.pi
        
    #METHOD
    def get_circumference(self):
        return self.radius*self.pi*2
```

```python
my_circle=Circle(30)
```

```python
my_circle.pi
```

```python
my_circle.radius
```

```python
my_circle.get_circumference()
```

```python
my_circle.area
```

```python
#THIS IS A BASE CLASS
class Animal():
    
    def __init__(self):
        print('ANIMAL CREATED')
        
    def who_am_i(self):
        print('I AM AN ANIMAL')
        
    def eat(self):
        print('I AM EATING')
```

```python
#THIS IS A DERIVED CLASS
class Dog(Animal):
    
    
    def __init__(self):
        Animal.__init__(self)
        print('Dog Created')
            
    def who_am_i(self):
        print('I am a dog!')
        
    def bark(self):
        print('WOOF!')
```

```python
mydog=Dog()
```

```python
mydog.eat()
```

```python
mydog.who_am_i()
```

```python
mydog.bark()
```

```python
my_animal=Animal()
```

```python
my_animal.eat()
```

```python
my_animal.who_am_i()
```

```python
class Dog():
    
    def __init__(self,name):
        self.name=name
        
    def speak(self):
        return self.name + ' says woof!'
```

```python
class Book():
    def __init__(self,title,author,pages):
        
        self.title=title
        self.author=author
        self.pages=pages
      
    def __str__(self):
        return f"{self.title} by {self.author}"
    
    def __len__(self):
        return self.pages
    
    def __del__(self):
        print('A book has been deleted')
```

```python
b=Book('Python rocks','Jose',200)
```

```python
print(b)
```

```python
str(b)
```

```python
len(b)
```

```python
del b
```
