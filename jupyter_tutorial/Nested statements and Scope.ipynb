```python
#GLOBAL
name='THIS IS A GLOBAL STRING'
def greet():
    
    #ENCLOSING FUNCTION
    name='Sammy'
    
    def hello():
        #LOCAL
        name='I m a local'
        print('Hello '+name)
        
    hello()
    
```

```python
greet()
```

```python
x= 50

def f(x):
    
    print(f'X is {x}')
    #LOCAL REASSIGNMENT ON A GLOBAL VARIABLE!
    x='NEW VALUE'
    print(f'I just locally changed X to {x}')
    return x
```

```python
print(x)
```

```python
x=f(x)
```
