```python
def check_even_list(num_list):
    #return all the even number in a list
    
    #placeholder variables
    even_numbers=[]
    for number in num_list:
        if number%2==0:
            even_numbers.append(number)
        else:
            pass
        
    return even_numbers
```

```python
check_even_list([1,6,5])
```

```python
def check_odd(num_list):
    odd_num=[]
    for num in num_list:
        if num%2!=0:
            odd_num.append(num)
        else:
            pass
    return odd_num
```

```python
work_hours = [('Abby',100),('Billy',1200),('Cassie',800)]
```

```python
work_hours
```

```python
def employee_check(work_hours):
    current_max=0
    current_employee=''
    for employee, hour in work_hours:
        if hour>current_max:
            current_max=hour
            current_employee=employee
        else:
            pass
    return current_employee,current_max
```

```python
employee_check(work_hours)
```
